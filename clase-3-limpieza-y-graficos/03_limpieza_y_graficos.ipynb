Las dos filas del hogar 42 son idénticas en las ocho columnas. Es un duplicado exacto y se saca en

# 3. Las tres formas de que no haya un dato

*Apunte, sección 4.1.*

En `ingreso` hay tres celdas que parecen decir lo mismo. Mirémoslas juntas.


````python
tres_casos = limpio[limpio["id"].isin([3, 38, 9])]
tres_casos[["id", "localidad", "personas", "ingreso"]]
````

El hogar 3 tiene el ingreso **vacío**, que pandas muestra como `NaN`. El hogar 38 tiene un **`0`**,
que es un ingreso declarado. Y el hogar 9 tiene **`999999`**, que es un centinela.

La celda que sigue mide qué le hace cada uno al promedio. Usa `.loc[filtro, columna]`, que es una
pieza nueva: el filtro elige las filas, igual que los corchetes de la clase 2, y el segundo término
se queda con una sola columna.


````python
todos = limpio["ingreso"].mean()
sin_centinela = limpio.loc[limpio["ingreso"] != 999999, "ingreso"].mean()
sin_el_cero = limpio.loc[(limpio["ingreso"] != 999999) & (limpio["ingreso"] != 0), "ingreso"].mean()

print("promedio con todo adentro:      ", round(todos))
print("promedio sin el centinela:      ", round(sin_centinela))
print("promedio sin el centinela ni el 0:", round(sin_el_cero))
````

El centinela sube el promedio unos veinte mil pesos, y **eso está mal**: no es un ingreso.

El cero lo baja unos seis mil, y **eso está bien**: es un hogar que declaró no tener ingreso.
Sacarlo sería mentir hacia arriba.

O sea: dos celdas que parecen iguales, y una hay que sacar y la otra hay que dejar. La diferencia no
está en el archivo: está en lo que significa cada una.


````python
limpio.loc[limpio["ingreso"] == 999999, "ingreso"] = None

print("ingresos declarados:", limpio["ingreso"].notna().sum(), "de", len(limpio))
print("faltantes por columna:")
print(limpio.isna().sum())
````

> **Esta celda falla a propósito.** Va a salir un bloque rojo y el notebook se frena ahí. Eso es lo

La última línea dice algo así:

```
ValueError: time data "2025-03-14" doesn't match format "%d/%m/%Y"
```

Es un error útil: te dice qué valor no encajó y con qué formato lo estabas comparando.

Y ahora la trampa de verdad. Hay una forma de que **no falle**, y es peor.


````python
peligrosa = pd.to_datetime(limpio["fecha_visita"], format="mixed")

print("convertidas:", peligrosa.notna().sum(), "de", len(limpio), " <- ninguna falla")
print("rango que da:", peligrosa.min().date(), "a", peligrosa.max().date())
print("fechas fuera de marzo y abril:", ((peligrosa.dt.month < 3) | (peligrosa.dt.month > 4)).sum())
````

Convirtió las treinta y siete, sin un solo error. Y **corrompió dos**.

`format="mixed"` adivina el formato de cada fecha por separado, y donde hay dos lecturas posibles
elige mes/día: `02/04/2025` se convirtió en el 4 de febrero y `07/04/2025` en el 4 de julio.
`25/03/2025` y `28/03/2025` se salvaron solas, porque no hay mes 25 ni mes 28. El relevamiento pasa
a ir de febrero a julio, cuando duró veintiséis días. Nadie lo avisa: hay que mirar el rango.

La forma correcta son dos pasos.


````python
fechas = pd.to_datetime(limpio["fecha_visita"], format="ISO8601", errors="coerce")
faltan = limpio.loc[fechas.isna(), "fecha_visita"]
fechas = fechas.fillna(pd.to_datetime(faltan, format="%d/%m/%Y"))

limpio["fecha_visita"] = fechas
print("convertidas:", fechas.notna().sum(), "de", len(limpio))
print("rango:", fechas.min().date(), "a", fechas.max().date())
print("duró", (fechas.max() - fechas.min()).days, "días")
````

Veintiséis días, del 14 de marzo al 9 de abril. Ese es el rango de verdad.

# 5. Extremos, codificación y escalado

*Apunte, secciones 5.4, 6.1 y 6.2.*

Los dos métodos, sobre la misma columna.


````python
ingreso = limpio["ingreso"].dropna()

q1, q3 = ingreso.quantile([0.25, 0.75])
iqr = q3 - q1
bajo, alto = q1 - 1.5 * iqr, q3 + 1.5 * iqr
z = (ingreso - ingreso.mean()) / ingreso.std()

print(f"Q1 = {q1:,.0f}   Q3 = {q3:,.0f}   IQR = {iqr:,.0f}")
print(f"límites del IQR: {bajo:,.0f} y {alto:,.0f}")
print(f"promedio = {ingreso.mean():,.0f}   desvío = {ingreso.std():,.0f}")
print()
print("marca el IQR:    ", sorted(ingreso[(ingreso < bajo) | (ingreso > alto)]))
print("marca el z-score:", sorted(ingreso[z.abs() > 2]))
````

Los dos coinciden en el `0` y en el `620000`, y **discrepan en el `310000`**, que es el hogar 7 de
la clase 2.

El z-score no lo marca porque el `620000` le agrandó el desvío. Con un desvío así, el `310000` queda
cerca del promedio. El IQR no usa el promedio, así que no le pasa eso.

Ninguno de los dos dice que el valor esté mal: los dos devuelven **candidatos**. Quien decide sos
vos.


````python
extremos = limpio[limpio["ingreso"].isin([0, 310000, 620000])]
extremos[["id", "localidad", "personas", "ingreso"]]
````

Y acá está el paso que ninguna función hace: **cruzar el extremo con el resto de su fila.**

El hogar 39 tiene 620.000 pesos y **doce personas**. Un hogar de doce personas con varios adultos
que aportan hace que ese ingreso sea plausible, no sospechoso. El hogar 38 declaró cero, y es un
dato.

Los tres se quedan. Lo que se anota es que se los revisó y por qué.
